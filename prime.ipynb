```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

for n in [2, 3, 5, 7, 11, 17, 20, 25, 29, 30]:
    if is_prime(n):
        print(n, "is prime")
    else:
        print(n, "is not prime")
```

Algorithm:

1. Take a number.
2. If the number is less than 2, it is not prime.
3. Check if it is divisible by any number from 2 to one less than itself.
4. If it is divisible, it is not prime.
5. Otherwise, it is prime.
